# Number Statistics Analyzer

Enter a list of numbers and let the program analyze them.

**It calculates:**
- Total numbers
- Sum
- Average
- Maximum / Minimum
- Even / Odd numbers
- Positive / Negative numbers / Zeros
- Count of each category

**Challenge:** build it with **functions** instead of one giant cell.

## Step 1 — Helper functions

Each category is extracted by its own small function. A list comprehension builds a new list from a condition in one line.

```python
def get_even_numbers(nums):
    return [n for n in nums if n % 2 == 0]


def get_odd_numbers(nums):
    return [n for n in nums if n % 2 != 0]


def get_positive_numbers(nums):
    return [n for n in nums if n > 0]


def get_negative_numbers(nums):
    return [n for n in nums if n < 0]


def get_zero_numbers(nums):
    return [n for n in nums if n == 0]
```

## Step 2 — The analyze function

`analyze_numbers` calls the helper functions and collects every result into one dictionary.

```python
def analyze_numbers(nums):
    stats = {}

    stats['total_numbers'] = len(nums)
    stats['sum'] = sum(nums)
    stats['average'] = sum(nums) / len(nums)

    stats['maximum'] = max(nums)
    stats['minimum'] = min(nums)

    stats['even'] = get_even_numbers(nums)
    stats['odd'] = get_odd_numbers(nums)

    stats['positive'] = get_positive_numbers(nums)
    stats['negative'] = get_negative_numbers(nums)
    stats['zeros'] = get_zero_numbers(nums)

    return stats
```

## Step 3 — Get input from the user

Ask for numbers as comma-separated values, split the string, strip whitespace, and convert each part to an integer.

```python
def get_user_input():
    raw = input('Enter numbers separated by commas: ').strip()

    numbers = []
    for part in raw.split(','):
        part = part.strip()
        if part != '':
            numbers.append(int(part))

    return numbers
```

## Step 4 — Show the results

The report is pretty-printed inside its own function. `main` runs the whole flow: input → analyze → display.

```python
def display_results(nums, stats):
    total = stats['total_numbers']
    total_sum = stats['sum']
    avg = stats['average']
    max_num = stats['maximum']
    min_num = stats['minimum']
    even = stats['even']
    odd = stats['odd']
    positive = stats['positive']
    negative = stats['negative']
    zeros = stats['zeros']

    print()
    print('========== NUMBER ANALYSIS ==========')
    print()
    print(f'Numbers: {nums}')
    print()
    print(f'Total Numbers : {total}')
    print(f'Sum           : {total_sum}')
    print(f'Average       : {avg}')
    print()
    print(f'Maximum       : {max_num}')
    print(f'Minimum       : {min_num}')
    print()
    print(f'Even Numbers  : {even}')
    print(f'Odd Numbers   : {odd}')
    print()
    print(f'Positive      : {positive}')
    print(f'Negative      : {negative}')
    print(f'Zeros         : {zeros}')
    print()
    print('Number of each category:')
    print(f'  Even count     : {len(even)}')
    print(f'  Odd count      : {len(odd)}')
    print(f'  Positive count : {len(positive)}')
    print(f'  Negative count : {len(negative)}')
    print(f'  Zero count     : {len(zeros)}')
    print()
    print('=====================================')


def main():
    numbers = get_user_input()

    if not numbers:
        print('No numbers entered. Exiting.')
        return

    stats = analyze_numbers(numbers)
    display_results(numbers, stats)
```

## Step 5 — Run it

The `if __name__ == '__main__':` guard lets us import the functions elsewhere without running the program automatically.

```python
if __name__ == '__main__':
    main()
```
